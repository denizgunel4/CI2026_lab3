```python
from random import Random


class Child:
    def __init__(self, name, characteristics):
        self.name = name
        self.characteristics = tuple(characteristics)

    def difference(self, other):
        return sum(a != b for a, b in zip(self.characteristics, other.characteristics))

    def __repr__(self):
        return self.name
```

```python
def fitness(solution):
    return sum(
        child.difference(solution[(i + 1) % len(solution)])
        for i, child in enumerate(solution)
    )

def tweak(solution, rng):
    first, second = rng.sample(range(len(solution)), 2)
    result = list(solution)
    result[first], result[second] = result[second], result[first]
    return result
```

```python
def evolutionary_algorithm(
    children,
    population_size=100,
    generations=500,
    elite_size=10,
    seed=2026,
):
    rng = Random(seed)
    children = list(children)
    population = [rng.sample(children, len(children)) for _ in range(population_size)]
    elite_size = min(elite_size, population_size)

    for _ in range(generations):
        population.sort(key=fitness, reverse=True)
        elites = population[:elite_size]
        population = elites + [
            tweak(rng.choice(elites), rng)
            for _ in range(population_size - elite_size)
        ]

    best = max(population, key=fitness)
    return best, fitness(best)
```

```python
NUM_CHILDREN = 20
NUM_CHARACTERISTICS = 20
rng = Random(42)

children = [
    Child(
        f"Child {i}",
        [rng.choice((False, True)) for _ in range(NUM_CHARACTERISTICS)],
    )
    for i in range(NUM_CHILDREN)
]

best_solution, best_score = evolutionary_algorithm(children)
print(f"Best score: {best_score}")
print(" -> ".join(child.name for child in best_solution))
```

```python
assert len(best_solution) == NUM_CHILDREN
assert set(best_solution) == set(children)
assert fitness(best_solution) == best_score
```
